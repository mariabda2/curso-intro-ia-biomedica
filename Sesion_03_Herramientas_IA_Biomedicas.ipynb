````python
# ============================================================
#  CELDA 2.1 — Consulta de PubMed con la API Entrez (Biopython)
#
#  Buscaremos artículos sobre IA en diagnóstico médico
#  y analizaremos las tendencias de publicación.
# ============================================================

Entrez.email = 'estudiante@universidad.edu'  # Identificación requerida por NCBI

termino_busqueda = 'artificial intelligence[Title/Abstract] AND diagnosis[Title/Abstract] AND 2020:2024[PDAT]'

print(f'🔍 Buscando en PubMed: "{termino_busqueda}"')
print('   Esto puede tomar unos segundos...')

try:
    handle = Entrez.esearch(db='pubmed', term=termino_busqueda, retmax=200)
    record = Entrez.read(handle)
    handle.close()

    total = int(record['Count'])
    ids   = record['IdList']

    print(f'\n📊 Resultados encontrados: {total:,} artículos')
    print(f'   Recuperados para análisis: {len(ids)}')

    # Recuperar resúmenes de los primeros 20 artículos
    handle2 = Entrez.efetch(db='pubmed', id=ids[:20], rettype='xml', retmode='xml')
    registros = Entrez.read(handle2)
    handle2.close()

    articulos = []
    for art in registros['PubmedArticle']:
        try:
            med = art['MedlineCitation']
            titulo = str(med['Article']['ArticleTitle'])
            try:
                abstract = str(med['Article']['Abstract']['AbstractText'][0])
            except:
                abstract = ''
            try:
                año = str(med['Article']['Journal']['JournalIssue']['PubDate'].get('Year', '?'))
            except:
                año = '?'
            articulos.append({'titulo': titulo[:80], 'año': año, 'abstract': abstract})
        except:
            pass

    df_pubmed = pd.DataFrame(articulos)
    print(f'\n📋 Muestra de artículos recuperados:')
    print(df_pubmed[['año', 'titulo']].head(10).to_string(index=False))

except Exception as e:
    print(f'⚠️  No se pudo conectar a NCBI (puede ser un problema de red en Colab).')
    print(f'   Simulando datos de ejemplo para continuar con el análisis...')

    # Datos simulados si NCBI no está disponible
    np.random.seed(42)
    años = np.repeat([2020, 2021, 2022, 2023, 2024], [3200, 4100, 5800, 7200, 4500])
    np.random.shuffle(años)
    df_pubmed = pd.DataFrame({'año': años[:200].astype(str)})
    df_pubmed['titulo'] = [f'IA en diagnóstico: estudio {i}' for i in range(200)]
    total = 24800
    print(f'   Datos simulados generados: {total:,} artículos totales (2020-2024)')
````

## 3. Procesamiento de Señales Biomédicas: Análisis de ECG

### 3.1 ¿Qué es una señal biomédica?

Una **señal biomédica** es cualquier medición física o química que registra un proceso biológico en función del tiempo. Las más comunes son:

| Señal | Origen | Herramienta de registro | Aplicaciones IA |
|-------|--------|------------------------|----------------|
| **ECG/EKG** | Actividad eléctrica cardíaca | Electrocardiógrafo | Detección de arritmias, infarto |
| **EEG** | Actividad neuronal cerebral | Electroencefalógrafo | Epilepsia, sueño, BCI |
| **EMG** | Actividad muscular | Electromiógrado | Rehabilitación, prótesis |
| **PPG** | Volumen sanguíneo (pulso) | Oxímetro / smartwatch | SpO₂, frecuencia cardíaca |
| **Respiración** | Movimiento torácico | Bandas torácicas | Apnea del sueño, EPOC |

### 3.2 El electrocardiograma (ECG)

El ECG registra la actividad eléctrica del corazón. Sus componentes principales son:

```
              ┌─ R (pico más alto)
              │
          Q ──┘  S──┐
   P─┐           │  └── ST ──── T─┐
     └───────────┘               └─
   │←── PR ──→│←──── QRS ────→│

  P:   Despolarización auricular (contracción auricular)
  QRS: Despolarización ventricular (contracción ventricular)
  T:   Repolarización ventricular (relajación)
```

La IA puede detectar en el ECG: **fibrilación auricular**, **bloqueos de rama**, **infarto de miocardio**, **arritmias ventriculares** y más, a veces con mayor precisión que un médico general.

## 4. Datos Ómicos: Reducción de Dimensionalidad

### 4.1 El problema de la alta dimensionalidad en biología

Los datos ómicos (genómica, transcriptómica, proteómica, metabolómica) tienen una característica desafiante: **muchas más variables que muestras**.

| Tipo de dato | Variables típicas | Muestras típicas |
|-------------|-------------------|------------------|
| Microarray de expresión génica | 20,000–50,000 genes | 50–500 muestras |
| RNA-seq (single-cell) | 30,000 genes | 1,000–100,000 células |
| Proteómica (MS) | 5,000–10,000 proteínas | 30–200 muestras |
| Metabolómica | 500–5,000 metabolitos | 50–300 muestras |

Para visualizar y analizar estos datos, usamos **reducción de dimensionalidad**: técnicas que comprimen miles de variables en 2 o 3 dimensiones manteniendo la estructura esencial.

### 4.2 Las tres técnicas más usadas

| Técnica | Tipo | Fortaleza | Limitación | Uso biomédico |
|---------|------|-----------|------------|---------------|
| **PCA** | Lineal | Rápido, interpretable, global | No captura relaciones no-lineales | Primera exploración, QC de datos |
| **t-SNE** | No lineal | Excelente para clusters | Lento, no preserva distancias globales | Visualización de scRNA-seq |
| **UMAP** | No lineal | Veloz, preserva estructura global | Hiperparámetros sensibles | scRNA-seq, proteómica, imagen |

````python
# ============================================================
#  CELDA 4.1 — Simular datos de expresión génica
#              (inspirado en RNA-seq de tumores)
#
#  Simulamos 4 tipos de tejido tumoral con perfiles de
#  expresión génica diferenciados, como en TCGA.
# ============================================================

np.random.seed(2024)

N_MUESTRAS = 240   # Biopsias totales
N_GENES    = 500   # Genes medidos
N_TIPOS    = 4     # Tipos tumorales

TIPOS_TUMORALES = [
    'Adenocarcinoma Pulmonar',
    'Carcinoma Basocelular',
    'Glioblastoma',
    'Carcinoma Colorrectal'
]

# Generar perfiles de expresión diferenciados por tipo tumoral
datos_expresion = []
etiquetas = []

for i, tipo in enumerate(TIPOS_TUMORALES):
    n = N_MUESTRAS // N_TIPOS
    # Cada tipo tiene un centroide diferente en el espacio génico
    centroide = np.zeros(N_GENES)
    # Genes marcadores diferenciales (bloques de genes sobreexpresados)
    bloque_inicio = i * 80
    bloque_fin    = bloque_inicio + 120
    centroide[bloque_inicio:bloque_fin] = np.random.uniform(2, 5, bloque_fin - bloque_inicio)

    # Ruido biológico
    muestras_tipo = centroide + np.random.normal(0, 0.8, (n, N_GENES))
    # Valores de expresión no negativos (log2-counts)
    muestras_tipo = np.clip(muestras_tipo, 0, None)

    datos_expresion.append(muestras_tipo)
    etiquetas.extend([tipo] * n)

X_omics = np.vstack(datos_expresion)
y_omics = np.array(etiquetas)

print(f'🧬 Dataset de expresión génica simulado:')
print(f'   {X_omics.shape[0]} biopsias × {X_omics.shape[1]} genes')
print(f'   Tipos tumorales: {N_TIPOS}')
print()
for tipo in TIPOS_TUMORALES:
    n = np.sum(y_omics == tipo)
    print(f'   • {tipo:<35} {n} muestras')
````

````python
# ============================================================
#  CELDA 4.2 — Comparación PCA vs t-SNE vs UMAP
# ============================================================

# Escalar datos
scaler_omics = StandardScaler()
X_scaled_omics = scaler_omics.fit_transform(X_omics)

# --- PCA ---
pca = PCA(n_components=50, random_state=42)
X_pca_50 = pca.fit_transform(X_scaled_omics)

pca2d = PCA(n_components=2, random_state=42)
X_pca2d = pca2d.fit_transform(X_scaled_omics)
varianza_exp = pca2d.explained_variance_ratio_ * 100

# --- t-SNE (sobre los primeros 50 PCs para eficiencia) ---
print('⏳ Calculando t-SNE...')
tsne = TSNE(n_components=2, perplexity=30, max_iter=1000,
            random_state=42, n_jobs=-1)
X_tsne = tsne.fit_transform(X_pca_50)

# --- UMAP ---
print('⏳ Calculando UMAP...')
try:
    import umap
    reducer = umap.UMAP(n_components=2, n_neighbors=20, min_dist=0.3,
                        random_state=42)
    X_umap = reducer.fit_transform(X_pca_50)
    umap_disponible = True
except:
    # Fallback si UMAP no cargó
    X_umap = X_tsne + np.random.normal(0, 2, X_tsne.shape)
    umap_disponible = False
    print('   (UMAP no disponible, usando aproximación)')

print('✅ Reducción de dimensionalidad completada.')

# Colormap por tipo tumoral
colores_tipos = {t: c for t, c in zip(TIPOS_TUMORALES, COLORES[:N_TIPOS])}
c_vector = [colores_tipos[t] for t in y_omics]

fig, axes = plt.subplots(1, 3, figsize=(19, 6))
fig.suptitle('Reducción de dimensionalidad en datos de expresión génica tumoral\n'
             f'(500 genes → 2 dimensiones, {N_MUESTRAS} biopsias)',
             fontsize=13, fontweight='bold')

datos_plots = [
    (X_pca2d,  f'PCA\n(PC1: {varianza_exp[0]:.1f}% | PC2: {varianza_exp[1]:.1f}% varianza)'),
    (X_tsne,   't-SNE\n(perplexity=30)'),
    (X_umap,   'UMAP\n(n_neighbors=20, min_dist=0.3)' if umap_disponible
                else 'UMAP (aprox.)\n(n_neighbors=20)')
]

for ax, (X_2d, titulo) in zip(axes, datos_plots):
    for tipo, color in colores_tipos.items():
        mask = y_omics == tipo
        ax.scatter(X_2d[mask, 0], X_2d[mask, 1],
                   c=color, s=35, alpha=0.75, edgecolors='white',
                   linewidth=0.4, label=tipo.split()[0])
    ax.set_title(titulo, fontsize=11, fontweight='bold')
    ax.set_xlabel('Componente 1')
    ax.set_ylabel('Componente 2')
    ax.legend(fontsize=8, title='Tipo tumoral')

plt.tight_layout()
plt.show()

print('💡 Observaciones:')
print('   PCA:   Separación lineal. Los clusters son menos compactos.')
print('   t-SNE: Clusters muy compactos y bien separados. Ideal para visualización.')
print('   UMAP:  Balance entre estructura local (clusters) y global (distancias).')
````

````python
# ============================================================
#  CELDA 4.3 — Scree Plot y clasificación sobre PCs
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# --- Scree Plot ---
varianza_acumulada = np.cumsum(pca.explained_variance_ratio_) * 100
axes[0].bar(range(1, 21), pca.explained_variance_ratio_[:20] * 100,
            color=C1, edgecolor='white', label='Varianza por PC')
ax_twin = axes[0].twinx()
ax_twin.plot(range(1, 21), varianza_acumulada[:20],
             color=C2, marker='o', linewidth=2, markersize=5,
             label='Varianza acumulada')
ax_twin.axhline(80, color='gray', linestyle='--', linewidth=1, alpha=0.7)
ax_twin.text(21, 81, '80%', color='gray', fontsize=9)
ax_twin.set_ylabel('Varianza acumulada (%)', color=C2)
ax_twin.set_ylim(0, 105)
axes[0].set_title('Scree Plot — Varianza explicada por PCA', fontsize=11, fontweight='bold')
axes[0].set_xlabel('Componente Principal')
axes[0].set_ylabel('Varianza explicada (%)')

# --- Clasificación con Random Forest sobre PCs ---
n_pcs_list = [5, 10, 20, 30, 50]
accuracies = []

for n_pcs in n_pcs_list:
    pca_n = PCA(n_components=n_pcs, random_state=42)
    X_pca_n = pca_n.fit_transform(X_scaled_omics)
    X_tr, X_te, y_tr, y_te = train_test_split(X_pca_n, y_omics,
                                               test_size=0.2, random_state=42,
                                               stratify=y_omics)
    rf = RandomForestClassifier(n_estimators=100, random_state=42)
    rf.fit(X_tr, y_tr)
    accuracies.append(rf.score(X_te, y_te))

axes[1].plot(n_pcs_list, [a*100 for a in accuracies],
             color=C4, marker='s', linewidth=2.5, markersize=9)
axes[1].fill_between(n_pcs_list, [a*100 for a in accuracies], alpha=0.15, color=C4)
for n, acc in zip(n_pcs_list, accuracies):
    axes[1].text(n, acc*100 + 0.5, f'{acc:.1%}', ha='center', fontsize=9, fontweight='bold')
axes[1].set_title('Exactitud de clasificación tumoral\nvs. número de PCs usados',
                  fontsize=11, fontweight='bold')
axes[1].set_xlabel('Número de componentes principales')
axes[1].set_ylabel('Exactitud (%)')
axes[1].set_ylim(80, 102)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
print('💡 Con solo 10-20 PCs (de 500 genes) se logra alta precisión:')
print('   PCA actúa como filtro de ruido y reduce el riesgo de sobreajuste.')
````

## 5. NLP Clínico: Extracción de Información de Notas Médicas

### 5.1 El tesoro escondido en el texto clínico

El **80% de la información médica** está en formato no estructurado: notas de evolución, informes de radiología, epicrisis, resúmenes de alta, resultados de patología. Durante décadas, este texto fue inutilizable para análisis a gran escala. El **NLP clínico** cambia eso.

### 5.2 Tareas de NLP en medicina

| Tarea NLP | Descripción | Ejemplo clínico |
|-----------|-------------|----------------|
| **Tokenización** | Dividir texto en unidades (palabras, oraciones) | Separar una nota de evolución en oraciones |
| **Named Entity Recognition (NER)** | Identificar entidades (síntomas, fármacos, diagnósticos) | Extraer «metformina 850mg» de una nota |
| **Relación entre entidades** | Identificar relaciones entre entidades | «paciente toma metformina para diabetes» |
| **Negación** | Detectar si una entidad está negada | «sin fiebre» ≠ «con fiebre» |
| **Clasificación de texto** | Asignar categorías a un documento | Clasificar nota como urgencia/rutina |
| **Resumen automático** | Condensar notas largas | Generar resumen de alta de una hospitalización |
| **Codificación ICD** | Asignar códigos diagnósticos automáticamente | CIE-10: E11.9 = Diabetes tipo 2 sin complicaciones |

### 5.3 Herramientas especializadas

- **spaCy + scispaCy**: NLP general + modelos biomédicos preentrenados
- **MedSpaCy**: Componentes para contexto clínico (negación, temporalidad, incertidumbre)
- **NLTK**: Toolkit clásico con recursos lingüísticos extensos
- **BioBERT / PubMedBERT**: Transformers preentrenados en literatura médica
- **cTAKES**: Pipeline de NLP clínico de la Clínica Mayo (Java)

````python
# ============================================================
#  CELDA 6.1 — Simulación de asistente clínico con LLM
#
#  Implementamos una versión simplificada de un asistente
#  de diagnóstico diferencial basado en síntomas.
#
#  ⚠️ IMPORTANTE: Esto es SOLO con fines educativos.
#  Un sistema de IA clínico real requiere validación
#  regulatoria exhaustiva antes de su uso clínico.
# ============================================================

# Base de conocimiento simplificada: síntomas → diagnósticos diferenciales
knowledge_base = {
    frozenset(['dolor pecho', 'diaforesis', 'disnea', 'irradiación']): {
        'diagnósticos': ['Síndrome Coronario Agudo', 'Angina Inestable',
                          'Disección Aórtica', 'Embolia Pulmonar'],
        'urgencia': '🔴 URGENCIA MÁXIMA',
        'acción': 'ECG inmediato + troponinas + activar protocolo STEMI/NSTEMI'
    },
    frozenset(['cefalea', 'fotofobia', 'náuseas']): {
        'diagnósticos': ['Migraña', 'Cefalea tensional', 'Meningitis',
                          'Hemorragia subaracnoidea'],
        'urgencia': '🟡 URGENCIA MODERADA',
        'acción': 'Evaluar rigidez nucal. Si hay signos meníngeos: TC + PL urgente'
    },
    frozenset(['fiebre', 'tos', 'disnea', 'expectoración']): {
        'diagnósticos': ['Neumonía bacteriana', 'COVID-19', 'Influenza',
                          'Exacerbación EPOC', 'Tuberculosis'],
        'urgencia': '🟡 URGENCIA MODERADA',
        'acción': 'Radiografía tórax + oximetría + hemocultivos si fiebre >38.5°C'
    },
    frozenset(['poliuria', 'polidipsia', 'pérdida de peso']): {
        'diagnósticos': ['Diabetes Mellitus tipo 1', 'Diabetes Mellitus tipo 2',
                          'Diabetes insípida'],
        'urgencia': '🟢 URGENCIA BAJA',
        'acción': 'Glucemia en ayunas + HbA1c + electrolitos'
    }
}

def asistente_diagnostico(sintomas_ingresados: list) -> dict:
    """Simulación simplificada de un sistema de diagnóstico diferencial."""
    sintomas_set = frozenset([s.lower().strip() for s in sintomas_ingresados])
    mejor_match = None
    mejor_score = 0

    for patron, info in knowledge_base.items():
        score = len(sintomas_set & patron) / len(patron)
        if score > mejor_score:
            mejor_score = score
            mejor_match = info
            mejor_patron = patron

    if mejor_match and mejor_score > 0.3:
        return {
            'sintomas_reconocidos': list(sintomas_set & mejor_patron),
            'diagnosticos_diferenciales': mejor_match['diagnósticos'],
            'nivel_urgencia': mejor_match['urgencia'],
            'accion_sugerida': mejor_match['acción'],
            'confianza': f'{mejor_score:.0%}'
        }
    return {'error': 'Síntomas no reconocidos. Consultar con médico.'}

# --- Casos de prueba ---
casos_prueba = [
    ['dolor pecho', 'diaforesis', 'irradiación'],
    ['fiebre', 'tos', 'expectoración', 'disnea'],
    ['cefalea', 'fotofobia', 'náuseas']
]

print('🤖 ASISTENTE DE DIAGNÓSTICO DIFERENCIAL (Simulación educativa)')
print('=' * 65)
print('⚠️  SOLO PARA FINES EDUCATIVOS — No reemplaza criterio clínico')
print('=' * 65)

for caso in casos_prueba:
    print(f'\n📋 Síntomas ingresados: {caso}')
    resultado = asistente_diagnostico(caso)
    if 'error' not in resultado:
        print(f'   {resultado["nivel_urgencia"]}')
        print(f'   Reconocidos:  {resultado["sintomas_reconocidos"]}')
        print(f'   Diagnósticos diferenciales:')
        for d in resultado['diagnosticos_diferenciales']:
            print(f'     • {d}')
        print(f'   Acción: {resultado["accion_sugerida"]}')
        print(f'   Confianza del match: {resultado["confianza"]}')
````

## 7. Pipeline Reproducible: Buenas Prácticas en IA Biomédica

### 7.1 ¿Por qué la reproducibilidad es una crisis en IA médica?

Un estudio publicado en *Nature Medicine* (2022) encontró que **solo el 47%** de los modelos de IA publicados en revistas médicas podían ser reproducidos por equipos independientes. Las causas principales:

- Código no publicado o incompleto
- Datos propietarios sin acceso
- Semillas aleatorias no fijadas
- Preprocesamiento no documentado
- Selección sesgada del conjunto de evaluación

### 7.2 Los 7 principios de un pipeline reproducible

| # | Principio | Implementación |
|---|-----------|---------------|
| 1 | **Semilla aleatoria fija** | `np.random.seed()`, `random_state=42` en todos los modelos |
| 2 | **Versionado de datos** | Hash del dataset, fecha de descarga |
| 3 | **Ambiente documentado** | `requirements.txt` con versiones exactas |
| 4 | **Separación estricta train/test** | El test set NO se toca hasta el final |
| 5 | **Validación cruzada** | Nunca evaluar solo en una partición |
| 6 | **Código versionado** | Git + GitHub/GitLab |
| 7 | **Reporte de incertidumbre** | Intervalos de confianza, no solo puntos |

````python
# ============================================================
#  CELDA 7.1 — Pipeline biomédico reproducible
#
#  Construimos un pipeline completo y documentado
#  usando el dataset de diabetes como caso de estudio.
# ============================================================

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.ensemble import GradientBoostingClassifier
import hashlib, json
from datetime import datetime

# ── CONFIGURACIÓN GLOBAL (reproducibilidad) ──────────────────
CONFIG = {
    'seed'          : 2024,
    'test_size'     : 0.20,
    'cv_folds'      : 5,
    'modelo'        : 'GradientBoostingClassifier',
    'fecha_ejecucion': datetime.now().strftime('%Y-%m-%d %H:%M:%S'),
    'dataset'       : 'Pima Indians Diabetes (NIDDK)',
    'objetivo'      : 'Predicción de diabetes tipo 2'
}
np.random.seed(CONFIG['seed'])

# ── 1. CARGA Y AUDITORÍA DE DATOS ────────────────────────────
url_db = ('https://raw.githubusercontent.com/jbrownlee/Datasets/'
           'master/pima-indians-diabetes.data.csv')
cols   = ['Embarazos','Glucosa','PresionArt','GrosorPiel',
           'Insulina','IMC','Pedigree','Edad','Diagnostico']
df_pipeline = pd.read_csv(url_db, names=cols)

# Hash del dataset para trazabilidad
hash_datos = hashlib.md5(df_pipeline.to_csv(index=False).encode()).hexdigest()[:12]
CONFIG['hash_dataset'] = hash_datos

print(f'📋 PIPELINE BIOMÉDICO REPRODUCIBLE')
print(f'   Dataset: {CONFIG["dataset"]}')
print(f'   Hash MD5: {hash_datos} (para verificar integridad)')
print(f'   Fecha: {CONFIG["fecha_ejecucion"]}')

# ── 2. PREPROCESAMIENTO DOCUMENTADO ──────────────────────────
vars_imputar = ['Glucosa','PresionArt','GrosorPiel','Insulina','IMC']
for v in vars_imputar:
    df_pipeline[v] = df_pipeline[v].replace(0, np.nan)

X_pipe = df_pipeline.drop('Diagnostico', axis=1)
y_pipe = df_pipeline['Diagnostico']

# ── 3. SEPARACIÓN TRAIN/TEST (el test NO se toca hasta el final) ──
X_tr, X_te, y_tr, y_te = train_test_split(
    X_pipe, y_pipe,
    test_size   = CONFIG['test_size'],
    random_state= CONFIG['seed'],
    stratify    = y_pipe
)
print(f'\n   Train: {len(X_tr)} | Test: {len(X_te)} (estratificado)')

# ── 4. PIPELINE SCIKIT-LEARN (impute → scale → model) ─────────
pipeline_final = Pipeline([
    ('imputar', SimpleImputer(strategy='median')),
    ('escalar', StandardScaler()),
    ('modelo' , GradientBoostingClassifier(
        n_estimators=150, max_depth=4,
        learning_rate=0.05, random_state=CONFIG['seed']
    ))
])

# ── 5. VALIDACIÓN CRUZADA (sobre TRAIN solamente) ─────────────
cv_strat = StratifiedKFold(n_splits=CONFIG['cv_folds'],
                            shuffle=True,
                            random_state=CONFIG['seed'])

cv_results = cross_validate(
    pipeline_final, X_tr, y_tr,
    cv=cv_strat,
    scoring=['roc_auc', 'f1', 'recall'],
    return_train_score=True,
    n_jobs=-1
)

print(f'\n📊 Validación cruzada ({CONFIG["cv_folds"]}-fold) en conjunto de entrenamiento:')
print(f'   ROC-AUC:     {cv_results["test_roc_auc"].mean():.3f} ± {cv_results["test_roc_auc"].std():.3f}')
print(f'   F1-Score:    {cv_results["test_f1"].mean():.3f} ± {cv_results["test_f1"].std():.3f}')
print(f'   Sensibilidad:{cv_results["test_recall"].mean():.3f} ± {cv_results["test_recall"].std():.3f}')

# ── 6. EVALUACIÓN FINAL EN TEST ────────────────────────────────
pipeline_final.fit(X_tr, y_tr)
y_te_pred = pipeline_final.predict(X_te)
y_te_prob = pipeline_final.predict_proba(X_te)[:, 1]

from sklearn.metrics import roc_auc_score, f1_score, recall_score
resultados_finales = {
    'AUC_test'          : round(roc_auc_score(y_te, y_te_prob), 4),
    'F1_test'           : round(f1_score(y_te, y_te_pred), 4),
    'Sensibilidad_test' : round(recall_score(y_te, y_te_pred), 4),
    'AUC_cv_media'      : round(cv_results['test_roc_auc'].mean(), 4),
    'AUC_cv_std'        : round(cv_results['test_roc_auc'].std(), 4)
}

print(f'\n🏆 Resultados FINALES en conjunto de TEST:')
for k, v in resultados_finales.items():
    print(f'   {k:<25}: {v}')

# ── 7. REPORTE DE CONFIGURACIÓN COMPLETO ──────────────────────
CONFIG.update(resultados_finales)
print(f'\n📁 Configuración completa (para reporte de reproducibilidad):')
print(json.dumps(CONFIG, indent=3))
````

### 8.2 La ética del dato en biomedicina

Cuando trabajas con datos biomédicos, debes hacerte siempre estas preguntas:

**Sobre los datos:**
- ¿Los pacientes dieron consentimiento informado para este uso de sus datos?
- ¿Los datos están anonimizados de forma irreversible?
- ¿Representan a la población a la que se aplicará el modelo?

**Sobre el modelo:**
- ¿Quién supervisa las predicciones antes de que afecten a un paciente?
- ¿Está el modelo validado externamente (en otra institución/país)?
- ¿Existe un mecanismo para reportar errores del modelo?

**Sobre el impacto:**
- ¿Beneficia por igual a todos los grupos demográficos?
- ¿Quién tiene acceso a la herramienta — solo hospitales ricos?
- ¿Cómo se actualiza el modelo cuando la epidemiología cambia?
